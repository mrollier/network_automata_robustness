```python
# check if the graph is directed
def is_directed(E:tc.Tensor):
    edge_list = set(map(tuple, E.numpy().T))
    for e in edge_list:
        if tuple(reversed(e)) not in edge_list:
            return True
    return False
```

```python
# check if the graph is directed
print(np.sum([ is_directed(E) for E, _ in NetworksDataset(path=DATA_PATH, cached=False, as_undirected=True) ]))
print(np.sum([ is_directed(E) for E, _ in NetworksDataset(path=DATA_PATH, cached=False, as_undirected=False) ]))
```
